```python
#finding outliers
```

```python
Q1=Q1 = df['Customer Age'].quantile(0.25)
Q3 = df['Customer Age'].quantile(0.75)
IQR = Q3 - Q1

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower bound: {Q1 - 1.5 * IQR}")
print(f"Upper bound: {Q3 + 1.5 * IQR}")

outliers = df[(df['Customer Age'] < Q1 - 1.5 * IQR) | (df['Customer Age'] > Q3 + 1.5 * IQR)]
print(f"\nOutliers found: {len(outliers)}")

outliers[['Customer Age', 'Units Sold']].head(10)
```

```python
#clean the outliers out
```

```python
upper_limit = Q3+ 1.5 * IQR
Lower_bound = Q1- 1.5 * IQR
df = df[(df['Customer Age']>= 18) & (df['Customer Age'] <= upper_limit)]
```

```python
#Create Charts 
#Histogram:How numerical data are distributed
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
plt.figure(figsize=(8, 5))
sns.histplot(df['Customer Age'], kde=True, color='skyblue')

plt.title('Distribution of Customer Age', fontsize=14)
plt.xlabel('Customer Age', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

```python
#How many individuals in each group 
```

```python
# 1. Define age bins and labels
bins = [18, 25, 35, 45, 55, 100]
labels = ['18-25', '26-35', '36-45', '46-55', '56+']

# 2. Create a new 'Age_Group' column in your dataframe
df['Age_Group'] = pd.cut(df['Customer Age'], bins=bins, labels=labels, right=True)

# 3. Count how many individuals are in each age group
age_group_counts = df['Age_Group'].value_counts().sort_index()
print(age_group_counts)
```
